# Buổi 4.3: Vision encoders và transfer learning
Freeze một encoder chưa pretrained có phải transfer learning tốt?

## Kiến thức
- Pretraining tạo weights đã học; architecture weights=None chưa có tri thức đó.
- Freeze giữ base; thay head/fine-tune khi dữ liệu và budget cho phép.
- ViT chia patch tokens, cần đúng preprocess/position theo checkpoint.

**Công thức:** Patch tokens=(H/P)×(W/P)


```python
# -*- coding: utf-8 -*-
import torch
from torch import nn
from torchvision.models import resnet18
torch.manual_seed(42)
torch.set_num_threads(1)

# weights=None minh họa architecture và freezing, KHÔNG phải pretrained model.
model=resnet18(weights=None)
for p in model.parameters(): p.requires_grad=False
model.fc=nn.Linear(model.fc.in_features,3)
model.eval()
X=torch.randn(2,3,64,64)
with torch.no_grad(): output=model(X)
print("Output shape:",tuple(output.shape))
print("Params cần train:",sum(p.numel() for p in model.parameters() if p.requires_grad))
print("Ví dụ này không đo chất lượng transfer learning.")
assert tuple(output.shape)==(2,3)
# Bài thực hành thật: dùng checkpoint được cấp trong nguồn và transforms đúng weights.
```
