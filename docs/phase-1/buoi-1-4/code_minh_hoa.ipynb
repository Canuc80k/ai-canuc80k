# Buổi 1.4: Data leakage và Pipeline
Nếu scaler biết trước mean của valid, điểm còn độc lập không?

## Kiến thức
- Preprocessing cũng học thống kê: median, mean/std, vocabulary, PCA.
- fit trên train; transform train/valid/test bằng cùng thống kê đã học.
- Pipeline trong CV giúp mỗi fold học preprocessing từ phần train của fold.

**Công thức:** fit(train) → transform(valid)


```python
# -*- coding: utf-8 -*-
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

train = np.array([[10.0],[np.nan],[30.0]])
valid = np.array([[np.nan],[100.0]])
# Tất cả thống kê được học từ train.
prep = make_pipeline(SimpleImputer(strategy="median"),StandardScaler())
prep.fit(train)
print("Median train:",prep[0].statistics_.tolist())
print("Mean sau impute train:",prep[1].mean_.tolist())
print("Valid sau impute:",prep[0].transform(valid).ravel().tolist())
print("Valid sau scale:",prep.transform(valid).round(4).ravel().tolist())
assert prep[0].statistics_[0] == 20
# Sai protocol: không gọi fit/fit_transform trên valid.
```

## Đọc kết quả
Pipeline chạy CV và dự đoán khu_vuc='Tay' theo bài gốc.
