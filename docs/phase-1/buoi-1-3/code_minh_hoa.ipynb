# Buổi 1.3: Chia dữ liệu và cross-validation
Ảnh cùng con vật vào cả train và valid có công bằng không?

## Kiến thức
- Train học tham số; validation chọn cấu hình; test đánh giá cuối.
- Stratified giữ tỉ lệ lớp; group split giữ đối tượng độc lập.
- Time split mô phỏng dự đoán tương lai; CV lặp protocol hợp lý qua folds.

**Công thức:** groups_train ∩ groups_valid = ∅


```python
# -*- coding: utf-8 -*-
import numpy as np
from sklearn.model_selection import GroupKFold,StratifiedKFold

# Hai ảnh cùng con vật phải thuộc cùng một group khi đánh giá cá thể mới.
ids = np.arange(8)
groups = np.array([0,0,1,1,2,2,3,3])
y = np.array([0,0,1,1,0,0,1,1])
X = ids.reshape(-1,1)
print("Group folds:")
for train,valid in GroupKFold(n_splits=2).split(X,y,groups):
    print("train",train.tolist(),"valid",valid.tolist())
    assert set(groups[train]).isdisjoint(set(groups[valid]))
print("Stratified giữ tỉ lệ lớp, không tự giữ group:")
for train,valid in StratifiedKFold(2,shuffle=True,random_state=42).split(X,y):
    overlap = sorted(set(groups[train]) & set(groups[valid]))
    print("groups trùng:",overlap)
```

## Đọc kết quả
Chỉ số split/folds và bảng CV; giải thích vì sao test không dùng để chọn.
