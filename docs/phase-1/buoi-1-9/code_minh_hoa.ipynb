# Buổi 1.9: Đạo hàm và gradient descent
Ta nên bước sang trái hay sang phải để giảm loss?

## Kiến thức
- Đạo hàm là độ dốc; gradient gom độ dốc của nhiều tham số.
- Gradient descent đi ngược gradient với độ dài bước do learning rate quyết định.
- Loss là thứ optimizer tối ưu; metric đánh giá có thể khác loss.

**Công thức:** w_new = w_old - learning_rate × gradient


```python
# -*- coding: utf-8 -*-

# f(w)=(w-3)^2, gradient=2*(w-3).
w = 0.0
learning_rate = 0.1
print("step | w | loss")
for step in range(5):
    loss = (w-3)**2
    print(step,round(w,4),round(loss,4))
    grad = 2*(w-3)
    w = w-learning_rate*grad
print("w sau 5 updates:",round(w,6))
assert abs(w-2.01696)<1e-8
```

## Đọc kết quả
Loss theo bước và cách cập nhật tham số từ bài sách.
