Một lỗi rất lớn ảnh hưởng MAE và RMSE giống nhau không?

## Kiến thức
- MAE lấy độ lớn sai số; MSE bình phương sai số, RMSE lấy căn sau khi trung bình.
- RMSE nhạy hơn với lỗi lớn; đơn vị RMSE giống target.
- Chọn và đọc metric của đề, không chọn metric thuận lợi cho mình.

**Công thức:** MAE=mean|ŷ-y|; RMSE=sqrt(mean(ŷ-y)²)


```python
# -*- coding: utf-8 -*-
import numpy as np

y = np.array([1.0,3.0])
pred = np.array([2.0,5.0])
error = pred-y
mae = np.abs(error).mean()
mse = (error**2).mean()
rmse = np.sqrt(mse)
print("Sai số:",error.tolist())
print("MAE:",mae)
print("MSE:",mse)
print("RMSE:",round(rmse,6))
assert mae==1.5 and mse==2.5
```
