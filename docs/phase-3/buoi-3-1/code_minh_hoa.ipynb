# Buổi 3.1: PyTorch tensors và autograd
Làm sao framework biết gradient của x*x+3*x?

## Kiến thức
- Tensor có shape, dtype, device; requires_grad cho phép theo dõi graph.
- backward dùng chain rule qua các phép toán đã ghi.
- grad được cộng dồn, cần zero_grad giữa các bước train.

**Công thức:** d(x²+3x)/dx = 2x+3


```python
# -*- coding: utf-8 -*-
import torch
torch.manual_seed(42)
torch.set_num_threads(1)

x = torch.tensor(2.0,requires_grad=True)
y = x*x+3*x
y.backward()
print("y:",y.item(),"dy/dx:",x.grad.item())
matrix = torch.arange(6,dtype=torch.float32).reshape(2,3)
print("shape:",tuple(matrix.shape),"dtype:",matrix.dtype,"device:",matrix.device)
assert x.grad.item()==7
# Numpy array không mang graph autograd của torch.
```
