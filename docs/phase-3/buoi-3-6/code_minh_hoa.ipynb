# Buổi 3.6: Optimizers, learning rate và initialization
Hai optimizer chạy cùng epochs có chắc cùng số updates không?

## Kiến thức
- SGD/mini-batch tính gradient từ batch; momentum giúp ổn định quỹ đạo.
- Adam/AdamW điều chỉnh update theo thống kê gradient; dùng đúng weight decay.
- So bằng cùng updates/budget và khởi tạo; chọn LR theo validation.

**Công thức:** SGD: w_new=w-lr×grad


```python
# -*- coding: utf-8 -*-
import torch
from torch import nn
torch.set_num_threads(1)

X = torch.linspace(-1,1,50).reshape(-1,1); y = 3*X-2
for name,optimizer_type in [("SGD",torch.optim.SGD),("AdamW",torch.optim.AdamW)]:
    torch.manual_seed(42)  # Hai lần bắt đầu cùng weights.
    model = nn.Linear(1,1)
    optimizer = optimizer_type(model.parameters(),lr=.1)
    for step in range(40):
        optimizer.zero_grad(); loss = ((model(X)-y)**2).mean()
        loss.backward(); optimizer.step()
    print(name,"loss",round(loss.item(),6),"w",round(model.weight.item(),4))
# Đây là so sánh 40 updates trên fixture, không khẳng định optimizer nào luôn tốt.
```
