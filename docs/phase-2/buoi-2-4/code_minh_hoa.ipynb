# Buổi 2.4: K-Nearest Neighbors
Đổi đơn vị một feature có làm đổi hàng xóm gần nhất không?

## Kiến thức
- kNN tìm k mẫu gần query và bỏ phiếu hoặc lấy trung bình.
- Khoảng cách phụ thuộc scale features; cần preprocessing học từ train.
- k nhỏ nhạy chi tiết; k lớn có thể làm mờ ranh giới.

**Công thức:** distance²(x,q)=sum_j(x_j-q_j)²


```python
# -*- coding: utf-8 -*-
import numpy as np
from sklearn.neighbors import KNeighborsClassifier

X = np.array([[0],[2],[10]],dtype=float)
y = np.array([0,0,1])
query = np.array([[3.0]])
for k in [1,3]:
    model = KNeighborsClassifier(n_neighbors=k).fit(X,y)
    distance,index = model.kneighbors(query)
    print("k",k,"neighbors",index[0].tolist(),"distance",distance[0].tolist())
    print("class:",int(model.predict(query)[0]))
assert KNeighborsClassifier(1).fit(X,y).predict(query)[0]==0
```
