# Buổi 2.3: Regularization L1/L2
Tại sao ta chấp nhận fit train kém hơn để hệ số bớt lớn?

## Kiến thức
- Regularization thêm penalty nhằm hạn chế lời giải quá phức tạp theo giả định.
- L1 khuyến khích nhiều hệ số 0; L2 khuyến khích hệ số nhỏ.
- Độ mạnh regularization chọn bằng validation; scaling thay ý nghĩa penalty.

**Công thức:** objective = data_loss + alpha × penalty


```python
# -*- coding: utf-8 -*-
import numpy as np
from sklearn.linear_model import LinearRegression,Ridge,Lasso

X = np.array([[0,0],[1,1],[2,2],[3,3],[4,4]],dtype=float)
y = np.array([0,2,4,6,8],dtype=float)
print("Hai features ở đây giống nhau hoàn toàn.")
for model in [LinearRegression(),Ridge(alpha=10),Lasso(alpha=0.5)]:
    model.fit(X,y)
    print(type(model).__name__,model.coef_.round(4).tolist())
print("L1([1,1]):",np.abs([1,1]).sum())
print("L2 squared([1,1]):",np.square([1,1]).sum())
# Không chọn penalty bằng điểm test.
```

## Hai bài trên lớp
1. Trả L1=sum|w| và L2_squared=sumw^2.
Input: `{"w": [1, -2, 3]}`
2. Không intercept: minimize sum(y-w*x)^2+alpha*w^2; w=sumxy/(sumx^2+alpha), denominator bằng 0→0.
Input: `{"x": [1, 2], "y": [2, 4], "alpha": 5}`
