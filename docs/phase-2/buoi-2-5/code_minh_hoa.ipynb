# Buổi 2.5: Decision Trees
Cây quyết định chọn câu hỏi ở nút đầu theo tiêu chí nào?

## Kiến thức
- Mỗi split chia dữ liệu; cây chọn giảm impurity/error.
- Classification lá dự đoán lớp; regression lá thường lấy trung bình target.
- Depth/min_samples_leaf hạn chế cây học quá chi tiết.

**Công thức:** Gini = 1 - sum_c p_c²


```python
# -*- coding: utf-8 -*-
from sklearn.datasets import make_moons
from sklearn.tree import DecisionTreeClassifier,export_text
from sklearn.model_selection import train_test_split

X,y = make_moons(n_samples=80,noise=.35,random_state=42)
Xtr,Xva,ytr,yva = train_test_split(X,y,test_size=.35,stratify=y,random_state=27)
for depth in [1,3,None]:
    model = DecisionTreeClassifier(max_depth=depth,random_state=42).fit(Xtr,ytr)
    print("depth",depth,"train",round(model.score(Xtr,ytr),4),
          "valid",round(model.score(Xva,yva),4))
    if depth==1:
        print(export_text(model,feature_names=["x1","x2"]).strip())
```

## Đọc kết quả
Cây và predictions; so train/validation, giải thích depth.
